## 1. 骨架版:三步公式(讲义 §2、§6)

每个分数取指数 → 求和 → 除以总和。

```python
import math

def softmax_naive(x):
    exp_x = [math.exp(v) for v in x]
    total = sum(exp_x)
    return [v / total for v in exp_x]

print('softmax([3,2,1]) =', [round(v, 4) for v in softmax_naive([3, 2, 1])])
print('三步手算对照: e^3, e^2, e^1 =', [round(math.exp(v), 4) for v in (3, 2, 1)],
      ' sum =', round(sum(math.exp(v) for v in (3, 2, 1)), 4))
```

## 2. 坑:溢出(讲义 §8)

分数一大,骨架版直接爆炸——作业 D1 的 NaN 就是它。

```python
try:
    softmax_naive([1000, 1001, 1002])
except OverflowError as e:
    print('OverflowError:', e)
```

## 3. 数值稳定版 + 平移不变性(讲义 §4、§9)

```python
def softmax(x):
    max_x = max(x)
    exp_x = [math.exp(v - max_x) for v in x]
    total = sum(exp_x)
    return [v / total for v in exp_x]

print('softmax([1000,1001,1002]) =', [round(v, 4) for v in softmax([1000, 1001, 1002])])
print('softmax([3,2,1])    =', [round(v, 4) for v in softmax([3, 2, 1])])
print('softmax([13,12,11]) =', [round(v, 4) for v in softmax([13, 12, 11])], ' <- 平移不变')
```

## 4. logits -> 概率:LLM 推理现场(讲义 §5、§12)

```python
vocab = ['我', '喜欢', '吃', '苹果', '香蕉']
logits = [1.2, 4.5, 2.1, 3.8, 0.5]
probs = softmax(logits)

for w, p in sorted(zip(vocab, probs), key=lambda t: -t[1]):
    print(f'{w}: {p:.1%}')
```
