```python
import numpy as np

def softmax_T(logits, T):
    x = np.asarray(logits) / T
    e = np.exp(x - x.max())
    return e / e.sum()

for T in (0.5, 1.0, 2.0, 10.0):
    print(f'T={T:5}:', np.round(softmax_T([3.0, 2.0, 1.0], T), 4))
```

```python
# 第一课 E1 的钩子兑现:整体放大 logits = 降温
from math import log
base = np.array([3., 2, 1])
print('softmax([3,2,1])      =', np.round(softmax_T(base, 1.0), 4))
print('softmax([30,20,10])   =', np.round(softmax_T(base * 10, 1.0), 4), '≈ T=0.1')
print('softmax([3,2,1], T=0.1) =', np.round(softmax_T(base, 0.1), 4))
```

```python
# T 扫描:第一名概率曲线(文本版)
for T in (0.1, 0.3, 0.5, 0.7, 1.0, 1.5, 2.0, 3.0, 5.0):
    p0 = softmax_T(base, T)[0]
    bar = '#' * int(p0 * 40)
    print(f'T={T:4}: p0={p0:.3f} {bar}')
```
